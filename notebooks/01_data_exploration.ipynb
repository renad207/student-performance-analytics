# 01 — Data Exploration & Cleaning
This notebook loads the synthetic 40,000-student dataset, checks quality, cleans it, and explores engagement vs performance.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel('../data/raw/student_performance_data.xlsx')
df.head()
```

```python
print(df.shape)
print(df.info())
print(df.isna().sum())
print(df.describe(include='all'))
```

```python
df['Video_Completion_Rate'] = df['Video_Completion_Rate'].fillna(df['Video_Completion_Rate'].median())
df['Study_Hours_Per_Week'] = df['Study_Hours_Per_Week'].fillna(df['Study_Hours_Per_Week'].median())
df['Percentage'] = df['Percentage'].clip(0,100)
df['Engagement_Score'] = (0.55*(df['App_Login_Count']/60*100)+0.45*df['Video_Completion_Rate']).round(2)
df['Engagement_Level'] = np.where(df['Engagement_Score']>=70,'High',np.where(df['Engagement_Score']>=45,'Medium','Low'))
df.to_csv('../data/processed/cleaned_data.csv',index=False)
df.to_csv('../data/processed/sql_ready_data.csv',index=False)
```

```python
print(df.groupby('Engagement_Level')['Percentage'].agg(['count','mean']).round(2))
print(df.groupby('Region')['Percentage'].mean().sort_values())
print(df.groupby('Subject')['Percentage'].mean().sort_values())
```

```python
plt.figure(figsize=(8,5)); plt.scatter(df['App_Login_Count'],df['Percentage'],alpha=.15); plt.xlabel('App Login Count'); plt.ylabel('Percentage'); plt.title('Logins vs Academic Performance'); plt.show()
plt.figure(figsize=(8,5)); plt.scatter(df['Video_Completion_Rate'],df['Percentage'],alpha=.15); plt.xlabel('Video Completion Rate'); plt.ylabel('Percentage'); plt.title('Video Completion vs Academic Performance'); plt.show()
```
