# 02 — SQL Analysis (SQLite)


```python
import sqlite3, pandas as pd
conn=sqlite3.connect(':memory:')
students=pd.read_csv('../data/processed/sql_ready_data.csv')
students.to_sql('students',conn,index=False,if_exists='replace')
```

```python
q='''SELECT Engagement_Level, COUNT(*) AS students, ROUND(AVG(Percentage),2) AS avg_percentage FROM students GROUP BY Engagement_Level ORDER BY avg_percentage DESC;'''
pd.read_sql_query(q,conn)
```

```python
q='''WITH regional AS (SELECT Region, AVG(Percentage) avg_pct FROM students GROUP BY Region) SELECT Region, ROUND(avg_pct,2) avg_pct, ROUND(avg_pct-(SELECT AVG(Percentage) FROM students),2) vs_overall FROM regional ORDER BY avg_pct DESC;'''
pd.read_sql_query(q,conn)
```

```python
q='''SELECT Student_ID, Region, Subject, App_Login_Count, Video_Completion_Rate, Percentage, Engagement_Level FROM students WHERE Engagement_Level='Low' AND Percentage < (SELECT AVG(Percentage) FROM students) ORDER BY Percentage ASC LIMIT 50;'''
pd.read_sql_query(q,conn)
```
