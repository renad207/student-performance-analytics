# 03 — Visualization Export
Creates stakeholder-ready summary CSVs for Power BI.


```python
import pandas as pd
df=pd.read_csv('../data/processed/cleaned_data.csv')
avg=df['Percentage'].mean()
df[(df['Engagement_Level']=='Low') & (df['Percentage']<avg)].to_csv('../data/processed/at_risk_students.csv',index=False)
df.groupby('Region',as_index=False).agg(Students=('Student_ID','count'),Avg_Percentage=('Percentage','mean'),Avg_Login_Count=('App_Login_Count','mean'),Avg_Video_Completion=('Video_Completion_Rate','mean')).round(2).to_csv('../data/processed/regional_summary.csv',index=False)
```
